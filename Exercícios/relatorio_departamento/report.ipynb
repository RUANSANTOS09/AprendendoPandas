```python
import pandas as pd

report = {
    'funcionario':['Ronald', 'Luiz', 'Fabio',
                   'Claudio', 'Laura', 'Fernanda'],
    'departamento':['Tecnologia', 'Tecnologia', 'Financeiro',
                    'Administrativo', 'RH', 'RH'],
    'salario':[6000, 12000, 14000, 9452, 2554, 1946]
}
df = pd.DataFrame(report)
df
```

```python
def faixa_salarial(salario):
    if salario <= 3000:
        return 'baixo'
    elif(salario >= 3000) and (salario <= 6000):
        return 'médio'
    else:
        return 'alto'
```

```python
dados_agrupados = df.groupby('departamento')['salario'].mean()
df_groupby = pd.DataFrame(dados_agrupados)
df_groupby
```

```python
df['faixa_salarial'] = df['salario'].apply(faixa_salarial)
df
```
