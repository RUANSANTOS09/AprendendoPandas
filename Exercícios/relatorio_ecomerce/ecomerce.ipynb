```python
import pandas as pd
report = {
    'produto':['Iphone 18 Pro Max', 'Iphone Duo', 'Mesa', 'Cheetos', 'Sofa', 'Iphone 16 Pro Max', 'Iphone 16 pro', 'Iphone 16 Pro Max', 'Iphone 18 Pro Max', 'Iphone Duo'],
    'categoria':['Tecnologia', 'Tecnologia', 'Movel', 'Alimento', 'Movel',
                 'Tecnologia', 'Tecnologia', 'Tecnologia', 'Tecnologia', 'Tecnologia'],
    'quantidade':[1, 1, 2, 10, 2, 1, 1, 1, 1, 1],
    'preco_unitario':[13000, 22000, 700, 12.90, 1500, 7000, 5000, 7589, 12000, 19886],
    'data_venda':['2026-01-10', '2026-01-13', '2026-01-30', '2026-12-01', '2026-11-30', '2026-05-02', '2026-07-08', '2026-06-05', '2026-09-20', '2026-09-28']

}
df = pd.DataFrame(report)
df
```

```python
df.info()
```

```python
df['valor_total'] = df['quantidade'] * df['preco_unitario']
df
```

```python
maior_faturamento = df[df['valor_total'] == df['valor_total'].max()]
maior_faturamento
```

```python
df_groupby = df.groupby(['categoria'])['valor_total'].sum()
df_groupby
```

```python
df['data_venda'] = pd.to_datetime(df['data_venda'], format='%Y-%m-%d')
```

```python
df['mes'] = df['data_venda'].dt.month
```

```python
maior_faturamento_mes = df.groupby(['mes'])['valor_total'].sum()
maior_faturamento_mes.idxmax()
```
