```python
import pandas as pd

report= {
    'produto': ['Samsung ZFold 8', 'Galaxy A57', 'Samsung S26 Ultra', 'Samsung ZFlip8', 'Galaxy A17'],
    'estoque_01':[8, 20, 5, 8, 25 ],
    'estoque_02':[5, 3, 2, 4, 10],
    'estoque_03':[6, 17, 3, 8, 15]
}

df = pd.DataFrame(report)
df
```

```python
df_melt = df.melt(
    id_vars=['produto'],
    value_vars=['estoque_01','estoque_02','estoque_03'],
    var_name='mes',
    value_name='quantidade_estoque'
)

df_melt
```

```python
df_melt[['estoque', 'mes']] = df_melt['mes'].str.rsplit('_', n=1, expand=True)
df_melt
```

```python
df_melt = df_melt.drop(columns=['estoque'])
df_melt
```

```python
df_melt.info()
```

```python
df_melt['mes'] = pd.to_datetime(df_melt['mes'], format='%m')
df_melt.info()
df_melt
```

```python
df_melt['mes'] = df_melt['mes'].dt.month
df_melt
```

```python
df_pivot_table = df_melt.pivot_table(
    index=['produto'],
    columns=['mes'],
    values=['quantidade_estoque'],
    aggfunc='first'
)
df_pivot_table
```
