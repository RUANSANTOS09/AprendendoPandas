```python
import pandas as pd

report = {
    'departamento':['Tecnologia', 'Recursos Humanos', 'Marketing',
                    'Financeiro', 'Operacoes', 'Administrativo'
                    ],
    'categoria_despesa':['Software e equipamentos', 'Salarios e benficios', 'Publicidade e campanhas', 'Taxas bancarias e servicos financeiros', 'Materiais e manutencao', 'Aluguel e despesas de escritorio'],
    'despesa_01-2024':[15000, 25000, 12000, 8000, 30000, 18000],
    'despesa_02-2024':[18500, 25500, 20000, 7500, 32500, 17800],
    'despesa_03-2024':[16200, 26000, 17500, 8300, 31000, 19200]

}
df = pd.DataFrame(report)
df
```

```python
df_melt = df.melt(
    id_vars=['departamento', 'categoria_despesa'],
    value_vars=['despesa_01-2024', 'despesa_02-2024', 'despesa_03-2024'],
    var_name='mes_despesa',
    value_name='valor_despesa'
)
df_melt
```

```python
df_melt.info()
```

```python
df_melt[['despesa', 'mes']] = df_melt['mes_despesa'].str.rsplit('_', n=1, expand=True)
```

```python
df_melt
```

```python
df_melt = df_melt.drop(columns=['mes_despesa', 'despesa'])
```

```python
df_melt
```

```python
df_melt['mes'] = pd.to_datetime(df_melt['mes'], format='%m-%Y')
df_melt
```

```python
df_melt.info()
```

```python
df_melt['mes'] = df_melt['mes'].dt.month_name(locale='pt_BR.UTF-8')
df_melt
```

```python
df_pivot_table = df_melt.pivot_table(
    index=['categoria_despesa'],
    columns=['mes'],
    values='valor_despesa',
    aggfunc='sum'
)

df_pivot_table
```
