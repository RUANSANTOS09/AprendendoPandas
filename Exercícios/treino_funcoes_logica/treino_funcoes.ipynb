```python
import pandas as pd
from pandas.core.dtypes import astype

df = pd.DataFrame({
    'register':[
        'João-25-SP',
        'Maria-17-RJ',
        'Pedro-32-MG',
        'Ana-8-BA',
        'Carlos-45-PR'
    ]
})
```

```python
df
```

```python
def classificar_pessoa(register):
    classification = register.split('-')
    name = classification[0]
    age = classification[1]
    state = classification[2]
    age = int(age)
    if age < 12:
        age = f'{name} é Criança ({state})'
    elif age >= 12 and age <= 17:
        age = f'{name} é Adolescente ({state})'
    else:
        age = f'{name} é Adulto ({state})'
    return age
```

```python
df['resumo'] = df['register'].apply(classificar_pessoa)
```
