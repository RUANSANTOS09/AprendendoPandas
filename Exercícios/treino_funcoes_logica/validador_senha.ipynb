```python
import pandas as pd
from pygments.lexers import special

df = pd.DataFrame({
    'senha': [
        'abc123',
        'Senha@2024',
        '12345678',
        'Ab1!',
        'MinhaSenh@@Forte1',
        'semNumero'
    ]
})

```

```python
def validar_senha(senha):
    problems = []
    special_characters = ['!', '@', '#', '$', '%', '^', '&', '*']
    tem_numero = False
    tem_maiuscula = False
    tem_special = False
    for i in senha:
        if i in special_characters:
            tem_special = True
        if i.isdigit():
            tem_numero = True
        if i.isupper():
            tem_maiuscula = True

    if len(senha) < 8:
        problems.append('Muito curta')
    if tem_numero == False:
        problems.append('Sem número')
    if tem_maiuscula == False:
        problems.append('Sem maiúscula')
    if tem_special == False:
        problems.append('Sem special')

    if not problems:
        return 'Senha forte'
    else:
        return ','.join(problems)

```

```python
df['resultado'] = df['senha'].apply(validar_senha)
```
