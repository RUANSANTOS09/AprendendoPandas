```python
import pandas as pd

notas = {
    'aluno': ['Pedro', 'Lucas', 'Luan', 'Felipe', 'Ricardo'],
    'nota_prova1':[6.8, 9.4, 8.2, 5.3, 7.8],
    'nota_prova2':[5.8, 8.2, 9.1, 7.2, 9.5],
    'nota_prova3':[8.4, 9.1, 7.4, 8.3, 9.8]
}
df = pd.DataFrame(notas)
df
```

```python
df_melt = pd.melt(
    df,
    id_vars=['aluno'],
    value_vars=['nota_prova1', 'nota_prova2', 'nota_prova3'],
    var_name='prova',
    value_name='notas'
)
```
