```python
import pandas as pd
from pandas.core.dtypes import astype

dados = {
    'produto_id':[1,2,3,4,5,6],
    'categoria':[
        'Eletronicos e Tecnologia',
        'Moda e Vestuário',
        'Casa e Decoração',
        'Alimentos e Bebidas',
        'Beleza e Cuidados Pessoais',
        'Esportes e Lazer'
    ],
    'faturamento_01-2024':[5000, 3000, 3000, 2500, 1000, 900],
    'faturamento_02-2024':[6000, 2000, 1000, 500, 2000, 1000],
    'faturamento_03-2024':[3000, 2500, 1500, 800, 1000, 700]
}
df = pd.DataFrame(dados)
```

```python
df
```

```python
df_melt = pd.melt(
    df,
    id_vars=['produto_id', 'categoria'],
        value_vars=[
        'faturamento_01-2024',
        'faturamento_02-2024',
        'faturamento_03-2024'
    ],
    var_name='mes',
    value_name='valor'
)
```

```python
df_melt
```

```python
df_melt[['mes', 'ano']] = df_melt['mes'].str.rsplit('_', n=1, expand=True)
```

```python
df_melt = df_melt.drop(columns=['mes'])
```

```python
df_melt
```

```python
df_melt[['mes', 'ano']] = df_melt['ano'].str.rsplit('-', expand=True)
```

```python
df_melt
```

```python
df_melt.info()
```

```python
df_melt['mes'] = pd.to_datetime(df_melt['mes'], format='%m')
```

```python
df_melt
```

```python
df_melt['mes'] = df_melt['mes'].dt.month
df_melt
```

```python
df_melt['ano'] = pd.to_datetime(df_melt['ano'], format='%Y')
df_melt['ano'] = df_melt['ano'].dt.year
```

```python
df_melt
```

```python
df_melt.info()
```

```python
df_melt['categoria'] = df_melt['categoria'].astype('category')
```

```python
df_melt.info()
```

```python
df_melt.groupby(['categoria', 'mes'])['valor'].sum()
```

```python
faturamento_esperado = 1000
filtro = df_melt[df_melt['valor'] < faturamento_esperado]
```
