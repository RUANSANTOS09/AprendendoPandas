```python
import pandas as pd

from Exercícios.pivot_table_exercicios.pivot_table import df_pivot_table

report = {
    'campanha_id':[1,2,3,4,5,6,7,8],
    'canal':['Instagram', 'Google Ads', 'Email', 'Email',
             'TikTok', 'Google Ads', 'Instagram', 'Twitter'],
    'investimento_01-2024':[4500, 8500, 1200, 1800,
                            3500, 6500, 4000, 1800
                            ],
    'investimento_02-2024':[4800, 9200, 1200, 1900,
                            3800, 7000, 4200, 1700
                            ],
    'investimento_03-2024':[5200, 8800, 1100, 2000,
                            4200, 7200, 4500, 1900
                            ],
    'cliques_01-2024':[3200, 4100, 1800, 2400,
                       2900, 3500, 3800, 1600
                       ],
    'cliques_02-2024':[3500, 4500, 1950, 2550,
                       3100, 3800, 4000, 1750
                       ],
    'cliques_03-2024':[3700, 4300, 1700, 2700,
                       3400, 3900, 4200, 1850
                       ]
}

df = pd.DataFrame(report)
df
```

```python
df_melt = df.melt(
    id_vars=['campanha_id', 'canal'],
    value_vars=[
        'investimento_01-2024','investimento_02-2024','investimento_03-2024', 'cliques_01-2024', 'cliques_02-2024', 'cliques_03-2024'
    ],
    var_name='metrica_mes',
    value_name='valor'
)
df_melt
```

```python
df_melt[['metrica','mes']] = df_melt['metrica_mes'].str.rsplit('_', n=1, expand=True)
df_melt
```

```python
df_melt = df_melt.drop(columns= 'metrica_mes')
```

```python
df_melt
```

```python
df_melt['mes'] = pd.to_datetime(df_melt['mes'], format='%m-%Y')
df_melt
```

```python
df_melt['mes'] = df_melt['mes'].dt.month
df_melt
```

```python
df_pivot_table = df_melt.pivot_table(
    index=['campanha_id', 'canal', 'mes'],
    columns=['metrica'],
    values=['valor'],
    aggfunc='first'
)
df_pivot_table
```

```python
df_pivot_table['custo_por_cliques'] = df_pivot_table['valor']['investimento'] / df_pivot_table['valor']['cliques']
df_pivot_table
```

```python
df_pivot_table['custo_por_cliques'] = df_pivot_table['custo_por_cliques'].round(2)
df_pivot_table
```

```python
custo_esperado = 1.50
filtro = df_pivot_table[df_pivot_table['custo_por_cliques'] > custo_esperado]
filtro
```
