```python
import pandas as pd

dados = {
    'produto':['Cama King','Sofa', 'Pc', 'Iphone 16 Pro', 'Fogão'],
    'preco':[2000,2500,5000,6000,2000],
    'quantidade':[4,5,3,2,10]
}
df = pd.DataFrame(dados)
df
```

```python
df['valor_total'] = df.apply(lambda x: x['preco'] * x['quantidade'] , axis=1)
```
