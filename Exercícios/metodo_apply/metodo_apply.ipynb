```python
import pandas as pd

dados = {
    'produtos':[
        'Iphone 17 Pro Max',
        'Iphone 17',
        'S25 FE',
        'Motorola Edge 60 Fusion',
        'Samsung Galaxy A57',
        'Xiaomi Poco X7 Pro',
        'Iphone 16',
        'Xiaomi 15T',
        'Edge 70 Pro'
    ],
    'preco':[10000, 5598, 3100, 3000, 2000, 2500, 5000, 3500, 4000]
}

df = pd.DataFrame(dados)
df
```

```python
def classificar_preco(preco):
    if preco <= 2000:
        return 'Barato'
    elif preco <= 3000:
        return 'Medio'
    else:
        return 'caro'
```

```python
df['faixa_preco'] = df['preco'].apply(classificar_preco)
```

```python
df
```

```python
df['preco_com_desconto'] = df['preco'].apply(lambda x: x - (x * 0.10))
```
