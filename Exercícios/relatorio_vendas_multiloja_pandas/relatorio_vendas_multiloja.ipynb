```python
import pandas as pd

dados = {
    'produto_id':[1,2,3,4,5,6],
    'loja':['Amazon', 'Amazon','Apple', 'Apple', 'Samsung', 'Samsung'],
    'produtos':['ALexa', 'Kindle', 'Iphone 17 Pro Max', 'Iphone 17 Pro', 'Samsung S24 Ultra', 'Samsung ZFLIP 7' ],
    'data_cadastro':[
        '01-02-2024', '01-30-2024',
        '02-10-2024', '01-22-2024',
        '03-12-2024', '03-22-2024'
    ],
    'venda_01-2024':[10912, 5832, 3482, 2365, 5978, 1784],
    'venda_02-2024':[9132, 2813, 2237, 3469, 3948, 884],
    'venda_03-2024':[8131, 3432, 1482, 2065, 4378, 314]
}

df = pd.DataFrame(dados)
df
```

```python
df_melt = pd.melt(
    df,
    id_vars=['produto_id', 'loja', 'data_cadastro', 'produtos'],
    value_vars=['venda_01-2024', 'venda_02-2024', 'venda_03-2024'],
    var_name='relatorio_mes',
    value_name='total_vendas'
)
df_melt
```

```python
df.info()
```

```python
df_melt['data_cadastro'] = pd.to_datetime(df_melt['data_cadastro'])
```

```python
df_melt[['relatorio','data_venda']] = df_melt['relatorio_mes'].str.rsplit('_', n=1, expand=True)
```

```python
df_melt
```

```python
df_melt = df_melt.drop(columns=['relatorio_mes', 'relatorio'])
```

```python
df_melt
```

```python
df_melt.info()
```

```python
df_melt['data_venda'] = pd.to_datetime(df_melt['data_venda'], format='%m/%Y')
```

```python
df_melt
```

```python
def classificar_produto(mes_cadastro, mes_venda):
    d1 = mes_cadastro.to_period('M')
    d2 = mes_venda.to_period('M')
    sub = d2 - d1
    if (sub.n >= 0) and (sub.n <= 3):
        return 'Novo'
    else:
        return 'veterano'
```

```python
df_melt['status_produto'] = df_melt.apply(lambda x: classificar_produto(x['data_cadastro'], x['data_venda']), axis=1)
```
