```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'seller':['Luiz ', 'pEdro','fernanda', 'fernanda'],
    'product':['Celular', 'Tv', 'Sofa', 'Sofa' ],
    'value':['R$ 1.200,00', '850.00', '4.500,00', '4.500,00'],
    'sale_date':['2020-15-29', '2025/09/10', '09-10-2025', '09-10-2025']
})
```

```python
df
```

```python
df['seller'] = (
    df['seller']
    .str.split()
    .str.join(" ")
    .str.title()
)
```

```python
def clear_price(value):
    if ',' in value:
        value = value.replace('.', '')  #1200,00
        value = value.replace(',', '.')  #1200.00
    if 'R$' in value:
        value = value.replace('R$', '')
    else:
        value = value
    return value
```

```python
df['value'] = df['value'].apply(clear_price)
df['sale_date'] = df['sale_date'].str.replace('/', '-')
```

```python
df.loc[df['sale_date'] == '09-10-2025', 'sale_date'] = '2025-10-09'
```

Conversão de número

```python
df['value'] = pd.to_numeric(df['value'], errors='coerce')
```

Conversão para DateTime

```python
df['sale_date'] = pd.to_datetime(df['sale_date'], format='%Y-%m-%d', errors='coerce')
```

```python
print(df['sale_date'].isna().sum())
```

```python
df.duplicated(subset=['seller', 'product', 'sale_date'], keep='last')
```
