```python
import pandas as pd
import mysql.connector as sql
```

```python
df['order_date'] = df['order_date'].str.strip()
```

```python
meses = {
    "Janeiro" : "01",
    "Fevereiro" : "02",
    "Março" : "03",
    "Abril" : "04",
    "Maio" : "05",
    "Junho" : "06",
    "Julho" : "07",
    "Agosto" : "08",
    "Setembro" : "09",
    "Outubro" : "10",
    "Novembro" : "11",
    "Dezembro" : "12"
}
for mes, numero in meses.items():
    df['order_date'] = df['order_date'].str.replace(mes,numero, regex=False)
```

```python
df['order_date'] = (
    df['order_date']
    .str.replace('.', '-', regex=False)
    .str.replace('/', '-', regex=False)
    .str.replace(' ', '-', regex=False)
)
```

```python
df['order_date'] = df['order_date'].str.strip('-')
```
