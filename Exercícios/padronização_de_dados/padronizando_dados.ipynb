```python
def normalization(value):
    if 'R$' in value  or '$' in value:
        value = value.replace('R$', ' ')
        value = value.replace('$', ' ')
    else:
        value = value
    return value
```

```python
def clear_price(value):
    if ',' in value:
        value = value.replace('.', '')
        value = value.replace(',', '.')
    elif len(value.split(".")[-1]) == 3:
        value = value.replace('.', '')
    else:
        value = value
    return value
```

```python
df['price'] = df['price'].apply(normalization)
df['price'] = df['price'].str.replace("O", "0")
df['price'] = df['price'].str.replace("-", " ")
df['price'] = df['price'].apply(clear_price)
df.loc[df['orders_raw_id'] == 14, 'price'] = '10500.00'
df['price'] = df['price'].str.strip()
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```
