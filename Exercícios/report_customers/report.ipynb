```python
import pandas as pd
import mysql.connector as sql
```

```python
print(df_customer.head(30))
```

```python
standardizing_info = ['first_name', 'last_name']
for column in standardizing_info:
    df_customer[column] = (
        df_customer[column]
        .str.split()
        .str.join(" ")
        .str.replace(' ', '',regex=False)
        .str.strip()
        .str.title()
    )
```

```python
df_customer['email'] = df_customer['email'].fillna('Email não cadastrado')
```

```python
df_order['order_date'] = df_order['order_date'].str.replace('/', '-')
```

```python
df = pd.merge(df_customer, df_order, on='customer_id')
```

```python
df.duplicated(subset=['amount', 'price', 'order_date', 'cpf'], keep='last')
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

```python
df = df.drop_duplicates(subset=['amount', 'price', 'order_date', 'cpf'], keep='last')
```
