```python
import pandas as pd
import mysql.connector as sql
```

```python
connection = sql.connect(
    host='localhost',
    user='root',
    password='Ruan81527733',
    database='restaurant'
)
df_orders = pd.read_sql("SELECT * FROM orders", connection)
df_customers = pd.read_sql("SELECT * FROM customer", connection)
```

```python
df_orders['order_date'] = df_orders['order_date'].str.replace('/', '-')
```

```python
standardizing_text = ['first_name', 'last_name']
for column in standardizing_text:
    df_customers[column] = (
        df_customers[column]
        .str.split()
        .str.join(" ")
        .str.replace(' ', '', regex=False)
        .str.strip()
        .str.title()
    )
```

```python
df_customers['email'] = df_customers['email'].fillna('Email não encontrado')
```

```python
df = pd.merge(df_orders, df_customers, on='customer_id')
```
