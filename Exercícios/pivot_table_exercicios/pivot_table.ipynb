```python
import pandas as pd

report = {
    'vendedor': ['Pedro', 'Pedro', 'Thomas', 'Peter', 'Freddy',
                 'Alexander', 'Ronald', 'Vicent', 'Vincent','Mariah'
                 ],
    'mes': [1, 1, 2, 3, 4, 5, 6, 7, 7, 8],
    'valor_vendido': [3451, 3716, 4629, 3528, 3281,
                      3971, 2869, 2918, 3989, 3018]
}
df = pd.DataFrame(report)
df
```

```python
df_pivot_table = df.pivot_table(
    index=['vendedor'],
    columns=['mes'],
    values=['valor_vendido'],
    aggfunc='sum'
)
df_pivot_table
```
