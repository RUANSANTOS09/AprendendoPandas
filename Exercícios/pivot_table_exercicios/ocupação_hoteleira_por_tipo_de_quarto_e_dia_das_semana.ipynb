```python
import pandas as pd

dados = {
    'tipo_quarto':['Standart', 'Standart', 'Solteiro', 'Duplo', 'Superior',
                   'Executivo', 'Master', 'Suite Junior', 'Villa', 'Villa'
                   ],
    'dia_semana':['Segunda', 'Segunda', 'Quarta', 'Quinta','Sexta','Sabado', 'Domingo', 'Domingo', 'Domingo', 'Domingo'],
    'reservas':[10, 8, 9, 14, 34, 38, 61, 56, 28, 34]
}
df = pd.DataFrame(dados)
df
```

```python
df_pivot_table = df.pivot_table(
    index=['tipo_quarto'],
    columns=['dia_semana'],
    values=['reservas'],
    aggfunc='mean'
)
df_pivot_table
```
