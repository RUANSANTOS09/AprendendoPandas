Metodo Pivot

```python
df_pivot = df.pivot(
    index=['Estudante'],
    columns=['Categoria', 'Semana', 'Tópico'],
    values=['Nota']
)
```

```python
df_pivot
```

```python
df_pivot.index
```

```python
df_pivot.columns
```

Stack


```python
df_pivot.stack(level=1)
```

Unstack

```python
df_pivot.stack(level=1).unstack()
```

Pivot Table

```python
df.pivot(
    index=['Estudante'],
    columns=['Categoria', 'Semana', 'Tópico'],
    values=['Nota']
)
```

```python
df.pivot(
    index=['Estudante'],
    columns=['Semana', 'Tópico'],
    values=['Nota']
)
```

```python

df.pivot_table(
    index = ['Estudante'],
    columns = ['Semana', 'Tópico'],
    values = ['Nota'],
    aggfunc='count'
)
```

```python
df.pivot_table(
    index = ['Estudante'],
    columns = ['Semana', 'Tópico'],
    values = ['Nota'],
    aggfunc='sum'
)
```

```python
df.pivot_table(
    index = ['Estudante'],
    columns = ['Semana', 'Tópico'],
    values = ['Nota']
)
```

```python
df.groupby(['Estudante', 'Semana', 'Tópico'])['Nota'].mean()
```

```python
df.groupby(['Estudante', 'Semana', 'Tópico'])['Nota'].mean().unstack(level=[1,2])
```
