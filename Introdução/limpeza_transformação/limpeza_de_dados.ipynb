## Criar outro dataframe para aplicar mais técnicas
### Leituras de temperaturas por um sensor, minuto a minuto, durante, meia hora

```python
temperaturas ={
    'Horas': ['00:00', '00:01', '00:02', '00:03', '00:04', '00:05', '00:06', '00:07', '00:08', '00:09', '00:10', '00:11', '00:12', '00:13', '00:14', '00:15', '00:16', '00:17', '00:18', '00:19', '00:20', '00:21', '00:22', '00:23', '00:24', '00:25', '00:26', '00:27', '00:28', '00:29', '00:30'],
    'Leitura': [14.5, 14.5, 14.6, 14.7, 14.6, 14.6, 14.7, 14.7, NaN, 14.8, 14.8, 14.8, 14.9, 14.9, 15.0, 15.0, NaN, 15.1, 15.1, 15.1, 15.2, 15.2, 15.2, 15.2, 15.3, 15.2, 15.3, 15.3, 15.3, NaN, 15.4]
}
df = pd.DataFrame(temperaturas)
df
```

```python
print(df.dtypes)
```

Verificar os dados nas linhas que possuem valores ausentes:

```python
ausentes = df[df.isna().any(axis=1)] # verificar dados True ao longo do eixe 1 (colunas)
print(ausentes)
```

Como tratar esses valores ausentes?
Podemos usar algumas abordagens, tais como:

1. Preencher com a média da coluna
2. Remover as linhas com NaN
3. Preencher com Métodos de Interpolação
4. Valores Anteriores ou Posteriores
5. Valores Fixos

1. Preencher com a média dos outros valores da coluna, com o método .fillna()

```python
media = df['Leitura'].mean()
df['Leitura'] = df['Leitura'].fillna(media)
print(df)
print(f'Média da temperatura: {media:.2f}')
```

```python
temperaturas ={
    'Horas': ['00:00', '00:01', '00:02', '00:03', '00:04', '00:05', '00:06', '00:07', '00:08', '00:09', '00:10', '00:11', '00:12', '00:13', '00:14', '00:15', '00:16', '00:17', '00:18', '00:19', '00:20', '00:21', '00:22', '00:23', '00:24', '00:25', '00:26', '00:27', '00:28', '00:29', '00:30'],
    'Leitura': [14.5, 14.5, 14.6, 14.7, 14.6, 14.6, 14.7, 14.7, NaN, 14.8, 14.8, 14.8, 14.9, 14.9, 15.0, 15.0, NaN, 15.1, 15.1, 15.1, 15.2, 15.2, 15.2, 15.2, 15.3, 15.2, 15.3, 15.3, 15.3, NaN, 15.4]
}
df = pd.DataFrame(temperaturas)
```

3. Preencher com Métodos de Interpolação (preenche com base em valores adjacentes)

```python
temperaturas ={
    'Horas': ['00:00', '00:01', '00:02', '00:03', '00:04', '00:05', '00:06', '00:07', '00:08', '00:09', '00:10', '00:11', '00:12', '00:13', '00:14', '00:15', '00:16', '00:17', '00:18', '00:19', '00:20', '00:21', '00:22', '00:23', '00:24', '00:25', '00:26', '00:27', '00:28', '00:29', '00:30'],
    'Leitura': [14.5, 14.5, 14.6, 14.7, 14.6, 14.6, 14.7, 14.7, NaN, 14.8, 14.8, 14.8, 14.9, 14.9, 15.0, 15.0, NaN, 15.1, 15.1, 15.1, 15.2, 15.2, 15.2, 15.2, 15.3, 15.2, 15.3, 15.3, 15.3, NaN, 15.4]
}
df = pd.DataFrame(temperaturas)
```

```python
df['Leitura'].interpolate(method='linear', inplace=True)
print(df)

# Mostrar a média de temperatura
media = df['Leitura'].mean()
print(f'Média da temperatura: {media:.2f}')
```

4. Preencher com Valores Anteriores ou Posteriores

```python
temperaturas ={
    'Horas': ['00:00', '00:01', '00:02', '00:03', '00:04', '00:05', '00:06', '00:07', '00:08', '00:09', '00:10', '00:11', '00:12', '00:13', '00:14', '00:15', '00:16', '00:17', '00:18', '00:19', '00:20', '00:21', '00:22', '00:23', '00:24', '00:25', '00:26', '00:27', '00:28', '00:29', '00:30'],
    'Leitura': [14.5, 14.5, 14.6, 14.7, 14.6, 14.6, 14.7, 14.7, NaN, 14.8, 14.8, 14.8, 14.9, 14.9, 15.0, 15.0, NaN, 15.1, 15.1, 15.1, 15.2, 15.2, 15.2, 15.2, 15.3, 15.2, 15.3, 15.3, 15.3, NaN, 15.4]
}
df = pd.DataFrame(temperaturas)
```

Preenchimento com valor anterior (forward fill)

```python
df['Leitura'] = df['Leitura'].ffill()
```

Preenchimento com valor posterior (backward fill)
df['Leitura'] = df['Leitura'].bfill()

```python
print(df)

# Mostrar a média de temperatura
media = df['Leitura'].mean()
print(f'Média da temperatura: {media:.2f}')
```

5. Preencher com Valores Fixos

```python
temperaturas ={
    'Horas': ['00:00', '00:01', '00:02', '00:03', '00:04', '00:05', '00:06', '00:07', '00:08', '00:09', '00:10', '00:11', '00:12', '00:13', '00:14', '00:15', '00:16', '00:17', '00:18', '00:19', '00:20', '00:21', '00:22', '00:23', '00:24', '00:25', '00:26', '00:27', '00:28', '00:29', '00:30'],
    'Leitura': [14.5, 14.5, 14.6, 14.7, 14.6, 14.6, 14.7, 14.7, NaN, 14.8, 14.8, 14.8, 14.9, 14.9, 15.0, 15.0, NaN, 15.1, 15.1, 15.1, 15.2, 15.2, 15.2, 15.2, 15.3, 15.2, 15.3, 15.3, 15.3, NaN, 15.4]
}
df = pd.DataFrame(temperaturas)
```
