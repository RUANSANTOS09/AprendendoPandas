```python
import pandas as pd
```

```python
base = './dados/dados_satisfacao.csv'
dados = pd.read_csv(base)
```

```python
dados
```

```python
dados_melt = pd.melt(
    dados,
    id_vars=['cliente_id', 'idade', 'genero'], # Colunas que não vamos mexer
    value_vars=[
        'satisfacao_geral_01-2024',
        'satisfacao_geral_02-2024',
        'satisfacao_produto_01-2024',
        'satisfacao_produto_02-2024',
        'satisfacao_produto_03-2024',
        'satisfacao_servico_01-2024',
        'satisfacao_servico_02-2024',
        'satisfacao_servico_03-2024'
    ], #Colunas de onde irão vir os nossos valores
    var_name='categoria_mes', #É o nome dessa nova coluna que tera como linhas as colunas de value_vars
    value_name='nota' #Nome da coluna onde ira vir os valores
)

dados_melt
```

Separando coluna categoria e mes

```python
#Exemplo de split
'Ruan Santos'.split()
```

```python
dados_melt[['categoria', 'mes']] = dados_melt['categoria_mes'].str.rsplit('_', n=1, expand=True)
```

```python
dados_melt
```

```python
dados_melt = dados_melt.drop(columns=['categoria_mes'])
```

Conversão mes

```python
dados_melt['mes'] = pd.to_datetime(dados_melt['mes'],format='%m-%Y')

dados_melt
```

```python
dados_melt.info()
```

```python
dados_melt['mes'] = dados_melt['mes'].dt.month

dados_melt
```

```python
dados_melt.info()
```

```python
dados_melt['genero'] = dados_melt['genero'].astype('category')
dados_melt['categoria'] = dados_melt['categoria'].astype('category')
```

```python
dados_melt.info()
```

```python
dados_melt.groupby(['categoria', 'mes'])['nota'].mean()
```
